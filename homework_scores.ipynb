```python
import pandas as pd
scores_data = {
    "Math": [80, 92, 75, 60, 88],
    "Physics": [70, 65, 90, 85, 78],
    "English": [95, 100, 85, 90, 87]
}
df_scores = pd.DataFrame(scores_data)
result_df4 = pd.DataFrame({
    "Середній бал": df_scores.mean().round(2),
    "Мінімальний бал": df_scores.min(),
    "Максимальний бал": df_scores.max()
})

best_subject = result_df4["Середній бал"].idxmax()

display(result_df4)
print(f"Предмет з найвищим середнім результатом: {best_subject}")
```
