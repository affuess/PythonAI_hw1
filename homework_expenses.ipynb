```python
import pandas as pd
expenses = [1200, 1500, 800, 600, 1000, 400, 900]
s_exp = pd.Series(expenses)
data_exp = {
    "Показник": [
        "Загальна сума витрат",
        "Середнє значення",
        "Максимальна витрата",
        "Мінімальна витрата",
        "Кількість днів з витратами > 1000"
    ],
    "Значення (грн)": [
        s_exp.sum(),
        round(s_exp.mean(), 2),
        s_exp.max(),
        s_exp.min(),
        (s_exp > 1000).sum()
    ]
}

result_df2 = pd.DataFrame(data_exp)
display(result_df2)
```
