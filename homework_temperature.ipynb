```python
%pip install pandas
import pandas as pd

temperatures = [2, 5, 4, 6, 7, 3, 1]
df_temp = pd.Series(temperatures)

data_temp = {
    "Показник": [
        "Середнє значення",
        "Максимальне значення",
        "Мінімальне значення",
        "Різниця (Max - Min)"
    ],
    "Значення (°C)": [
        round(df_temp.mean(), 2),
        df_temp.max(),
        df_temp.min(),
        df_temp.max() - df_temp.min()
    ]
}

result_df1 = pd.DataFrame(data_temp)
display(result_df1)
```
